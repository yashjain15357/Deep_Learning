# **Classification Loss Functions**

## **Binary Cross-Entropy (BCE) Loss**
**Binary Cross-Entropy (BCE) Loss**, jise **Log Loss** bhi kehte hain, binary classification problems (jahan target output sirf do categories mein hota hai: $0$ ya $1$, jaise Spam/Not Spam) ke liye standard loss function hai.

Ye function predict ki gayi probability ($\hat{y}$) ko actual label ($y$) se compare karta hai aur galat predictions par high penalty lagata hai.

---

### **Formula**

$$BCE = -\frac{1}{n} \sum_{i=1}^{n} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

Where:

* $n$ = total number of samples / data points
* $y_i$ = actual binary class label ($0$ ya $1$)
* $\hat{y}_i$ = predicted probability of class 1 ($0$ se $1$ ke beech, e.g., Sigmoid activation ka output)
* $\log$ = natural logarithm ($\ln$)

---

### **Example Data**

Maano humare paas $n = 3$ samples hain jinki actual labels aur predicted probabilities niche di gayi hain:

| Sample ($i$) | Actual Label ($y_i$) | Predicted Probability ($\hat{y}_i$) |
| --- | --- | --- |
| **1** | $1$ (Spam) | $0.8$ (High confidence - Correct) |
| **2** | $0$ (Not Spam) | $0.2$ (Low confidence for Class 1 - Correct) |
| **3** | $1$ (Spam) | $0.3$ (Low confidence for Class 1 - Wrong) |

---

### **Step-by-Step Calculation**

Har sample ke liye formula ka ye part evaluate hota hai: $L_i = -\left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$

#### **Step 1: Sample 1 ke liye Loss Calculate Karein ($y_1 = 1, \hat{y}_1 = 0.8$)**

Since $y_1 = 1$, second term $(1 - y_1)$ zero ho jayega:


$$L_1 = -\left[ 1 \cdot \log(0.8) + (1 - 1) \cdot \log(1 - 0.8) \right]$$

$$L_1 = -\log(0.8) \approx -(-0.223) = 0.223$$

---

#### **Step 2: Sample 2 ke liye Loss Calculate Karein ($y_2 = 0, \hat{y}_2 = 0.2$)**

Since $y_2 = 0$, first term $y_2 \log(\hat{y}_2)$ zero ho jayega:


$$L_2 = -\left[ 0 \cdot \log(0.2) + (1 - 0) \cdot \log(1 - 0.2) \right]$$

$$L_2 = -\log(0.8) \approx -(-0.223) = 0.223$$

---

#### **Step 3: Sample 3 ke liye Loss Calculate Karein ($y_3 = 1, \hat{y}_3 = 0.3$)**

Since $y_3 = 1$, second term zero ho jayega:


$$L_3 = -\left[ 1 \cdot \log(0.3) + (1 - 1) \cdot \log(1 - 0.3) \right]$$

$$L_3 = -\log(0.3) \approx -(-1.204) = 1.204$$


*(Note: Galat prediction ke karan yahan loss kaafi high ($1.204$) aaya hai.)*

---

#### **Step 4: Overall Average Loss Calculate Karein**

Teeno samples ke losses ka average lete hain:

$$BCE = \frac{L_1 + L_2 + L_3}{3}$$

$$BCE = \frac{0.223 + 0.223 + 1.204}{3} = \frac{1.650}{3} = 0.550$$

**Final Loss:** **$0.550$**

---

### **Key Takeaways**

* **Confident & Correct:** Jab prediction sahi aur confident ho (jaise Sample 1 & 2), loss kam aata hai ($0.223$).
* **Wrong Prediction:** Jab model galat predict karta hai (jaise Sample 3 me actual $1$ tha par prediction $0.3$), loss steep ho jata hai ($1.204$).
* **Log Penalty:** Logarithm ki waja se jaise-jaise probability actual value se door jaati hai, penalty exponentially badhti hai.

## **Softmax Activation Function**
**Softmax Activation Function** ka main kaam multi-class classification neural networks ke **raw outputs (logits)** ko **probabilities** me convert karna hai.

Jab aapka neural network final layer par output deta hai, toh woh real numbers hote hain (jaise $2.5$, $-1.2$, $0.8$), jinhe **Logits** kaha jata hai. Softmax in logits ko pick karke $0$ se $1$ ke beech scale karta hai, aur sabhi classes ki probabilities ka sum hamesha **1 (100%)** hota hai.

---

### Softmax Formula

Formula standard mathematical format me yeh hota hai:

$$S(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{C} e^{z_j}}$$

* **$z_i$**: Current class ka raw logit score.
* **$e^{z_i}$**: Exponential function, jo negative values ko positive banata hai aur bde numbers ko high boost deta hai.
* **$\sum e^{z_j}$**: Sabhi classes ke exponential values ka total sum (normalization ke liye).

---

### Step-by-Step Example

Maano aapka model 3 classes (**Cat, Dog, Frog**) analyze kar raha hai aur final layer par ye logits deta hai:

* $z = [2.0, 1.0, 0.1]$

**1. Exponentials Calculate Karein ($e^z$):**

* $e^{2.0} \approx 7.389$
* $e^{1.0} \approx 2.718$
* $e^{0.1} \approx 1.105$

**2. Sum Calculate Karein:**

* $\text{Sum} = 7.389 + 2.718 + 1.105 = 11.212$

**3. Softmax Probabilities ($S(z_i)$):**

* **Cat:** $\frac{7.389}{11.212} \approx \mathbf{0.659}$ ($65.9\%$)
* **Dog:** $\frac{2.718}{11.212} \approx \mathbf{0.242}$ ($24.2\%$)
* **Frog:** $\frac{1.105}{11.212} \approx \mathbf{0.099}$ ($9.9\%$)

---

### Key Highlights

* **Probability Distribution:** Neural network ka predicted output directly class confidence show karta hai.
* **Loss Function Integration:** Categorical Cross-Entropy loss function ke saath easily integrate hoke smooth gradients aur better training optimize karta hai.
* **Sigmoid vs Softmax:** Sigmoid Binary Classification (2 classes) ke liye use hota hai, jabki Softmax Multi-Class Classification (> 2 classes) ke liye standard method hai.

## **Sparse Categorical Crossentropy (SCCE)**
**Sparse Categorical Crossentropy (SCCE)** multi-class classification problems me ek popular **loss function** hai. Jab aapka model multiple classes (jaise 0 to 9 digits, ya Cat/Dog/Bird) predict karta hai, toh yeh loss function measure karta hai ki model ka predicted output actual target se kitna alag hai.

---

### Main Role and Features

* **Integer Target Representation:** Iska sabse bada role yeh hai ki yeh targets ko **integer labels** ($0, 1, 2, \dots$) ke roop me accept karta hai. Aapko manual [One-Hot Encoding](https://www.geeksforgeeks.org/deep-learning/sparse-categorical-crossentropy-vs-categorical-crossentropy/) (jaise $[0, 1, 0]$) karne ki zaroorat nahi padti.
* **Memory & Computational Efficiency:** High-dimensional data ya bohot saari classes (e.g., 1000+ categories) wale datasets me [One-Hot Encoding](https://www.geeksforgeeks.org/deep-learning/sparse-categorical-crossentropy-vs-categorical-crossentropy/) memory consume karti hai. SCCE internally label handling karke memory save karta hai aur training speed badhata hai.
* **Optimization Guidance:** Training ke dauran model ke predictions ko actual label se compare karke loss score compute karta hai, jisse gradient descent optimizer weights ko update kar sake.

---

### Mathematical Formula

Calculation ke liye SCCE internal level par correct class index ki predicted probability ka **Negative Logarithm** nikalta hai:

$$L = -\log(\hat{y}_k)$$

* **$\hat{y}_k$**: Target class $k$ ki predicted probability (jo Softmax layer se milti hai).

---

### Step-by-Step Working Example

Maano aap [MNIST Handwritten Digit Classification](https://www.geeksforgeeks.org/deep-learning/categorical-cross-entropy-in-multi-class-classification/) par kaam kar rahe hain, jahan classes $0$ se $9$ hain:

1. **Target Label:** An image has true class $y = 3$ (Integer format).
2. **Model Output ([Softmax](https://www.geeksforgeeks.org/deep-learning/categorical-cross-entropy-in-multi-class-classification/)):** Model 10 probabilities generate karta hai. Maano class 3 ke liye prediction score $\hat{y}_3 = 0.8$ ($80\%$ confidence) hai.
3. **Loss Calculation:**

$$\text{Loss} = -\log(0.8) \approx 0.223$$


4. **Interpretation:** Agar model sahi class ke liye kam confidence ($\hat{y}_3 = 0.2$) deta, toh loss high hota ($-\log(0.2) \approx 1.609$). Loss function high error par penalty deta hai taaki model accurate ho sake.

---

### SCCE vs Categorical Crossentropy

| Feature | Sparse Categorical Crossentropy | Categorical Crossentropy |
| --- | --- | --- |
| **Label Format** | Integers (e.g., `3`) | One-Hot Encoded vectors (e.g., `[0, 0, 0, 1, 0]`) |
| **Memory Requirement** | Kam (Efficient) | Jyada (Large matrices) |
| **Best For** | Jab targets already integer indices hon | Jab probability distributions/soft targets use ho rahe hon |

Model building ke time, output layer par **Softmax activation function** ke saath SCCE loss function ka use multi-class tasks me optimal performance deta hai.
