## **Vanishing and Exploding Gradients Problems in Deep Learning**

Deep Learning models (khaaskar deep networks aur RNNs) me training ke dauran **Vanishing** aur **Exploding Gradients** do sabse badi dikkaten hain, jo **Backpropagation** ke time aati hain.

---

### Step 1: Gradient aur Backpropagation kya hai?

Neural network me hum loss (galti) ko kam karne ke liye weights ko update karte hain using **Gradient Descent**:

$$\text{Weight New} = \text{Weight Old} - (\text{Learning Rate} \times \text{Gradient})$$

* **Gradient** batata hai ki loss ke respect me weight ko kitna aur kis direction me badalna hai.
* Backpropagation me hum final output layer se peeche (earlier layers) ki taraf **Chain Rule** ka use karke gradients calculate karte hain:

$$\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a_n} \times \frac{\partial a_n}{\partial a_{n-1}} \times \dots \times \frac{\partial a_1}{\partial w_1}$$

Jab network bahot deep hota hai, toh yeh repeated multiplication samasya khadi karta hai.

---

### Step 2: Vanishing Gradient Problem

* **Problem:** Jab backpropagation karte waqt gradients repeated multiplication ki wajah se itne chhote ($< 1$) ho jaate hain ki zero ke paas pahunch jaate hain.
* **Kyu hota hai?**
* Sigmoid ya Tanh jaise activation functions ka derivative hamesha $0.25$ ya usse chhota hota hai.
* Deep network me jab aise chhotey numbers bar-bar multiply hote hain (jaise $0.25 \times 0.25 \times 0.25 \dots$), toh final value lagbhag $0$ ho jaati hai.


* **Impact:** Earlier (pehli) layers ke weights update hi nahi hote, yaani network ki shuruaati layers seekhna (learn karna) band kar deti hain.

---

### Step 3: Exploding Gradient Problem

* **Problem:** Jab gradients repeated multiplication ki wajah se exponentially badhte jaate hain aur bahot bade ($> 1$) ho jaate hain.
* **Kyu hota hai?**
* Agar initial weights ya derivatives 1 se bade hote hain (jaise $2 \times 2 \times 2 \dots$), toh deep networks me gradient value $1000$ ya infinity ($\infty$) tak chali jaati hai.


* **Impact:** Weights me massive (bade) updates hote hain, jisse loss curve destabilize ho jata hai aur model diverge kar jata hai (`NaN` errors aane lagte hain).

---

### Step 4: Inko Solve Kaise Karein? (Solutions)

* **ReLU Activation Function:** Sigmoid/Tanh ki jagah **ReLU** (Rectified Linear Unit) ya Leaky ReLU ka use karein, kyunki positive values ke liye iska derivative $1$ rehta hai.
* **Proper Weight Initialization:** Weights ko sahi tarike se initialize karein, jaise **Xavier / Glorot** (Sigmoid ke liye) ya **Kaiming / He Initialization** (ReLU ke liye).
* **Batch Normalization:** Layers ke inputs ko normalize karta hai jisse training stable rehti hai.
* **Gradient Clipping:** Exploding gradients ko rokne ke liye gradient ki maximum value par ek limit (threshold) laga di jaati hai.
* **ResNet / Gated Architectures:** Residual networks (Skip connections) aur RNNs ke case me **LSTM / GRU** ka use karein jo long-term gradients ko float hone me help karte hain.
