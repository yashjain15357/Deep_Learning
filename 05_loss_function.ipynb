**Loss Function** AI/Machine Learning model ka ek crucial part hota hai jo ye measure karta hai ki aapka model predictions kitne accurate kar raha hai.

Aasan bhasha mein: Loss function batata hai ki target value (actual answer) aur model prediction (predicted answer) ke beech mein kitna **error (duri/difference)** hai.

---

### **Loss Function ka Use Kya Hai?**

* **Error Measure Karna:** Ye calculate karta hai ki prediction actual output se kitna galat hai.
* **Model ko Guide Karna:** Optimization algorithms (jaise Gradient Descent) loss function ke value ko use karke model ke **weights aur biases** ko update karte hain.
* **Loss Value Zero/Low Karna:** Loss jitna kam (lower) hoga, model ki performance aur accuracy utni hi acchi hogi.

---

### **Types of Loss Functions**

Loss functions ko mostly dho main tasks ke basis par divide kiya jata hai: **Regression** aur **Classification**.

#### **1. Regression Loss Functions** (Jab output continuous numerical value ho, e.g., Price, Age prediction)

* **Mean Squared Error (MSE):**
* Actual value aur predicted value ke difference ka square calculate karke uska average leta hai.
* **Formula:** $MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$
* **Use:** Sabse common hai, lekin outliers ke prati bohot sensitive hota hai kyunki squaring error ko bada bana deti hai.


* **Mean Absolute Error (MAE):**
* Actual aur predicted value ke absolute difference ka average leta hai.
* **Formula:** $MAE = \frac{1}{n} \sum_{i=1}^{n} \vert{}y_i - \hat{y}_i\vert{}$
* **Use:** Outliers ke case mein MSE se zyada robust/stable hota hai.


* **Huber Loss:**
* MSE aur MAE ka combination hai. Small errors ke liye MSE use karta hai aur large errors/outliers ke liye MAE.



---

#### **2. Classification Loss Functions** (Jab output categories/classes ho, e.g., Spam/Not Spam, Cat/Dog)

* **Binary Cross-Entropy Loss (Log Loss):**
* Binary classification (2 classes - 0 ya 1) ke liye use hota hai.
* **Formula:** $L = - \frac{1}{n} \sum_{i=1}^{n} [y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i)]$
* **Use:** Jab output class probability (0 se 1 ke beech) ho.


* **Categorical Cross-Entropy Loss:**
* Multi-class classification (3 ya usse zyada classes) ke liye use hota hai jab labels one-hot encoded hon.


* **Sparse Categorical Cross-Entropy Loss:**
* Categorical Cross-Entropy jaisa hi hai, lekin yahan class labels integers (0, 1, 2...) ke roop mein hote hain, non-one-hot.


* **Hinge Loss:**
* Mainly Support Vector Machines (SVMs) mein use hota hai binary classification ke liye, jahan labels $-1$ aur $1$ hote hain.
