## **weather the person exited the bank or not**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , LabelEncoder,OneHotEncoder
import pickle
```

```python
data = pd.read_csv('../dataset/Churn_Modelling.csv')
data.head()
```

```python
data.info()
```

```python
# preprocessing the data
# drop the irrelecant columns
data = data.drop(['RowNumber','CustomerId','Surname'] , axis=1)
data
```

```python
# gender and geographical region has categorical feature so convert in the int
label_encoder_gender = LabelEncoder()
data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
```

```python
data['Geography'].unique()
```

```python
# apply OneHot encoding on the Geography column
onehot_encoder_geography = OneHotEncoder(
    sparse_output=False
)
geography_arr = onehot_encoder_geography.fit_transform(data[['Geography']])
geography_col = onehot_encoder_geography.get_feature_names_out(["Geography"])
print(geography_col)
```

```python
geo_data =pd.DataFrame(geography_arr , columns=geography_col)
geo_data
data = pd.concat([data.drop('Geography' , axis=1) , geo_data],axis=1)
data
```

```python
# save the label encoder pickel file
with open('label_encoder_gender.pkl' , 'wb') as file:
    pickle.dump(label_encoder_gender , file)
```

```python
# save one hot encoder pickel file
with open('onehot_encoder_geography.pkl' , 'wb') as file:
    pickle.dump(onehot_encoder_geography ,  file)
```

```python
# divide the dataset into the dependent or independetn feature
x = data.drop('Exited' , axis = 1)
y= data['Exited']

# split data into train test 
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)

# scale those feature
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test  = scaler.transform(x_test)
```

```python
x_train
```

```python
with open('scaler.pkl' , 'wb') as file:
    pickle.dump(scaler , file)
```

# **ANN Implementation**

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense , Input
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
model = Sequential([
    Input(shape = (x_train.shape[1],)),
    Dense(64 , activation='relu' ), #hl1 connected with input layer
    Dense(32 , activation='relu'), #hl2
    Dense(1, activation='sigmoid') #output layer
])
```

```python
model.summary()
```

```python
opt = tf.keras.optimizers.Adam(learning_rate=0.01)
# loss = tf.keras.losses.binary_crossentropy()
```

```python
model.compile(optimizer=opt , loss='binary_crossentropy' , metrics=['accuracy'])
```

```python
# setup tensorboard
from tensorflow.keras.callbacks import EarlyStopping , TensorBoard
log_dir = 'logs/fit/' + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback = TensorBoard(log_dir= log_dir , histogram_freq=1)
```

```python
# setup Early EarlyStopping
early_stopping_callback = EarlyStopping(monitor='val_loss' , patience=15 , restore_best_weights=True)
```

```python
# train the model
history = model.fit (
    x_train , 
    y_train , 
    validation_data=(x_test , y_test),
    epochs = 100, 
    callbacks = [tensorflow_callback , early_stopping_callback]
)
```

```python
model.save('model.h5')
```
