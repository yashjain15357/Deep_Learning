```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np

```

```python
# load the train midel , scaler pickle , onehot
model = load_model('model.h5')

# load encoder and scaler
with open('label_encoder_gender.pkl' , 'rb')as file:
    label_encoder_gender = pickle.load(file)

with open('onehot_encoder_geography.pkl' , 'rb')as file:
    onehot_encoder_geography = pickle.load(file)

with open('scaler.pkl' , 'rb')as file:
    scaler = pickle.load(file)
```

```python
input_data = {
    "CreditScore": 539,
    "Geography": "France",
    "Gender": "Male",
    "Age": 37,
    "Tenure": 2,
    "Balance": 127609.59,
    "NumOfProducts": 1,
    "HasCrCard": 1,
    "IsActiveMember": 0,
    "EstimatedSalary": 98646.22
}

```

```python
geo_encoded = onehot_encoder_geography.transform([[input_data['Geography']]])
geo_encoded_df = pd.DataFrame(geo_encoded , columns=onehot_encoder_geography.get_feature_names_out(['Geography']))
geo_encoded_df


```

```python
input_data = pd.DataFrame([input_data])
input_data=pd.concat([input_data.reset_index(drop=True) , geo_encoded_df] , axis=1)
input_data
```

```python
input_data['Gender'] = label_encoder_gender.transform(input_data['Gender'])
```

```python
input_data.drop(['Geography'] , axis=1 , inplace=True)
input_data
```

```python
# scaling the input data
input_scale = scaler.transform(input_data)
input_scale
```

```python
prediction = model.predict(input_scale)
prediction
```

```python
prediction_probablity = prediction[0]
prediction_probablity
```

```python
if prediction_probablity > 0.5:
    print('the customer is likely to chur.')

else:
    print('The customer is not likely to churn')
```
