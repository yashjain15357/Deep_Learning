## **Regression Loss Functions**
Regression Loss Functions un Machine Learning / Deep Learning models mein use hote hain jahan target variable continuous numerical value hota hai (jaise house price, temperature, ya crop yield predict karna). Ye functions actual values ($y$) aur predicted values ($\hat{y}$) ke beech ke error ko measure karte hain.

## **Mean Squared Error (MSE)**
**Mean Squared Error (MSE)**, jise **L2 Loss** bhi kehte hain, deep learning aur machine learning mein sabse widely used regression loss function hai. Ye model ki predicted output ($\hat{y}$) aur actual target output ($y$) ke beech ke squared differences ka average calculate karta hai.

---

### **Formula**

$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

Where:

* $n$ = total data points / batch size
* $y_i$ = actual value (true label)
* $\hat{y}_i$ = predicted value (model output)

---

### **Step-by-Step Calculation**

Maano aapke paas $n = 3$ samples hain jahan aap house prices predict kar rahe hain:

* **Step 1: Calculate Error (Residual)**
Har data point ke liye actual value se predicted value ko subtract karke difference nikalte hain.

$$\text{Error}_i = y_i - \hat{y}_i$$


* **Step 2: Square the Error**
Har error value ko square karte hain. Isse:
1. Negative error values positive ho jaati hain (taaki positive aur negative errors ek dusre ko cancel na karein).
2. Bade errors ko extra penalty milti hai.

$$\text{Squared Error}_i = (y_i - \hat{y}_i)^2$$




* **Step 3: Sum All Squared Errors**
Sare data points ke squared errors ko add kar lete hain.

$$\text{Sum} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$


* **Step 4: Take the Average**
Total sum ko number of data points ($n$) se divide karke average loss nikalte hain.

$$MSE = \frac{1}{n} \text{Sum}$$



---

### **Advantages of MSE**

* **Mathematical Smoothness (Differentiable):** MSE ka derivative (gradient) har jagah continuous aur smooth hota hai. Is wajah se **Gradient Descent** aur **Backpropagation** ke waqt weights ko update karna kafi aasan aur fast hota hai.
* **Penalizes Large Errors:** Square karne ki wajah se chote errors ki tulna mein bade errors par zyada penalty lagti hai. Ye model ko force karta hai ki wo bade mistakes avoid kare.
* **Convex Function:** MSE loss curve convex (bowl-shaped) hota hai, jisse optimization algorithm ko global minimum dhoondhne mein aasani hoti hai.

---

### **Disadvantages of MSE**

* **Sensitive to Outliers:** Agar dataset mein outliers (extreme value mistakes) hain, toh squaring hone ke kaaran MSE value bohot zyada badh jaati hai. Isse gradient explode ho sakta hai aur model main data ki jagah outliers par zyada focus karne lagta hai.
* **Unit Mismatch:** Since errors ko square kiya jata hai, MSE ki unit original data ki unit se alag hoti hai (e.g., agar price Rupees mein hai, toh MSE $Rupees^2$ hoga). Isliye interpretation ke liye hum **RMSE (Root MSE)** ka use karte hain.
* **Vanishing Gradient Issue:** Neural networks mein deep layers ya linear activations ke sath jab prediction actual value se bohot door hoti hai, toh gradients extreme hone par model convergence unstable ho sakta hai.

## **Mean Absolute Error (MAE)**
**Mean Absolute Error (MAE)**, jise **L1 Loss** bhi kehte hain, deep learning aur machine learning mein regression tasks ke liye use hone wala ek essential loss function hai. Ye predicted outputs ($\hat{y}$) aur actual target values ($y$) ke beech ke **absolute differences** ka average measure karta hai.

---

### **Formula**

$$MAE = \frac{1}{n} \sum_{i=1}^{n} \vert{}y_i - \hat{y}_i\vert{}$$

Where:

* $n$ = total number of data points / batch size
* $y_i$ = actual value (ground truth)
* $\hat{y}_i$ = predicted value (model output)
* $\vert{} \cdot \vert{}$ = absolute value operator (jo distance ko hamesha positive banata hai)

---

### **Step-by-Step Calculation**

Maano aapke paas $n = 3$ samples hain aur aap house prices predict kar rahe hain:

* **Step 1: Calculate Difference (Error)**
Har data point ke liye actual value se predicted value ko subtract karein:

$$\text{Difference}_i = y_i - \hat{y}_i$$


* **Step 2: Take Absolute Value**
Difference ka absolute value nikalen taaki sign ignore ho jaye aur bas magnitude (distance) bache:

$$\text{Absolute Error}_i = \vert{}y_i - \hat{y}_i\vert{}$$


* **Step 3: Sum All Absolute Errors**
Sabhi samples ke absolute errors ko add karein:

$$\text{Total Error} = \sum_{i=1}^{n} \vert{}y_i - \hat{y}_i\vert{}$$


* **Step 4: Calculate Average**
Total sum ko $n$ (sample count) se divide karein:

$$MAE = \frac{1}{n} \times \text{Total Error}$$



---

### **Advantages of MAE**

* **Robust to Outliers:** Outliers (extreme incorrect values) ke hone par MAE unhe MSE ki tarah square nahi karta. Is wajah se overall loss explode nahi hota aur model outliers se distrub hone se bachtaa hai.
* **Easy Interpretation:** MAE ki unit same wahi hoti hai jo original target variable ki unit hoti hai (e.g., agar target variable Rupees mein hai, toh MAE error bhi Rupees mein hi milega).
* **Linear Penalty:** Har error magnitude ke saath linearly badhta hai, jo ki steady aur balanced training optimization deta hai.

---

### **Disadvantages of MAE**

* **Non-Differentiable at Zero:** $y_i = \hat{y}_i$ (yaani error = 0) par absolute value function ka gradient (derivative) undefined hota hai. Gradient descent optimization mein is wajah se minor instability aa sakti hai.
* **Constant Gradients:** Gradient constant रहता hai ($\pm 1$), chhah-e error bada ho ya bohot chota. Isse jab model global minimum ke paas pahunchta hai, tab gradients automatically small nahi hote, jisse exact minimum converge karna mushkil ho sakta hai.
* **No Extra Penalty for Large Errors:** Baaz aukat models ke liye bade mistakes par zyada penalty lagana zaruri hota hai (jaise autonomous driving ya safety systems mein), jahan MAE MSE ki tarah harsh penalty nahi laga pata.

## **Huber Loss**
**Huber Loss** (jise Smooth L1 Loss bhi kehte hain) ek hybrid regression loss function hai jo **Mean Squared Error (MSE)** aur **Mean Absolute Error (MAE)** dono ke best properties ko combine karta hai.

Ye small errors ke liye MSE (quadratic) ki tarah act karta hai aur large errors ke liye MAE (linear) ki tarah act karta hai.

---

### **Formula**

$$L_\delta(y, \hat{y}) = \begin{cases}  \frac{1}{2}(y - \hat{y})^2 & \text{for } \vert{}y - \hat{y}\vert{} \le \delta \\ \delta \vert{}y - \hat{y}\vert{} - \frac{1}{2}\delta^2 & \text{for } \vert{}y - \hat{y}\vert{} > \delta  \end{cases}$$

Where:

* $y$ = Actual target value
* $\hat{y}$ = Predicted value
* $\delta$ (Delta) = Threshold parameter jo decide karta hai ki kab loss MSE se MAE par switch hoga.

---

### **Advantages of Huber Loss**

* **Robust to Outliers:** Large errors ($\vert{}y - \hat{y}\vert{} > \delta$) ke liye ye linear (MAE) behavior apply karta hai. Isse outliers ke karan loss explode nahi hota aur model stable rehta hai.
* **Differentiable at Zero:** MAE zero par non-differentiable hota hai, lekin Huber Loss zero aur baaki sabhi points par differentiable hai. Is wajah se **Gradient Descent** smooth convergence deta hai.
* **Best of Both Worlds:**
* Small errors ke liye MSE ki tarah **fast convergence** aur **smooth gradients**.
* Large errors ke liye MAE ki tarah **outlier tolerance**.



---

### **Disadvantages of Huber Loss**

* **Hyperparameter Tuning ($\delta$):** Threshold parameter $\delta$ ko manually tune karna padta hai. Sahi $\delta$ value set na karne par model suboptimal performance de sakta hai.
* **Computationally Expensive:** Formula mein condition check ($\vert{}y - \hat{y}\vert{} \le \delta$) hone ki wajah se MSE ya MAE ki tulna mein iski computational overhead thodi zyada hoti hai.
* **Complex Implementation:** Standard MSE/MAE ke muqable iski mathematical equation piece-wise hoti hai, jisse custom implementations thodi complex ho sakti hain.
