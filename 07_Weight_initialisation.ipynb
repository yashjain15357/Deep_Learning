Yeh article **Piyush Kashyap** ka likha hua hai, jisme unhone neural networks me **Weight Initialization** (weights ko shuruat me kya value deni chahiye) ke baare me detail me bataya hai.

Is pure article ka aasaan aur detailed breakdown neeche diya gaya hai:

---

### **1. Weight Initialization Kyun Zaroori Hai?**

Jab aap koi neural network train karte ho, toh sabse pehle weights ko ek initial value deni padti hai. Agar yeh values sahi nahi chunoge, toh model sahi se seekh nahi payega.

**Galat initialization se yeh problems aati hain:**

* **Vanishing Gradient Problem:** Gradients itne chhote ho jaate hain ki network ke pehle waale layers update hi nahi hote, aur training rukh jaati hai.
* **Exploding Gradient Problem:** Gradients itne bade ho jaate hain ki model unstable ho jaata hai aur values infinity tak chali jaati hain.
* **Slow Convergence:** Model ko optimal solution tak pahunchne me bahut zyaada time lagta hai.

---

### **2. Weight Initialization Ke Types**

#### **1. Zero Initialization**

Is method me sare weights ko **0** set kar diya jaata hai.

* **Problem:** Yeh tarika bilkul kaam nahi karta kyunki sare neurons ek hi jaisa behave karne lagte hain (symmetry problem). Unhe alag-alag features seekhne ka mauka nahi milta.

#### **2. Random Initialization**

Isme weights ko randomly choose kiya jaata hai, par isme bhi do bade issues aate hain:

* **Chhote Random Weights:** Agar weights 0 ke bahut paas chunoge, toh **vanishing gradient** ki dikkat aayegi aur training bohot slow ho jayegi.
* **Bade Random Weights:** Agar weights bohot bade chunoge, toh **exploding gradient** ki dikkat aayegi jisse model unstable ho jayega.

#### **3. Xavier / Glorot Initialization**

Yeh method har layer me activations ki variance ko balance me rakhta hai. Isme weights ek normal distribution se liye jaate hain jiska variance layer ke input/output neurons par depend karta hai.

* **Best For:** **Sigmoid** aur **Tanh** activation functions ke saath.
* **Benefit:** Vanishing aur exploding gradients dono ko rokta hai.

#### **4. He Initialization**

Yeh Xavier initialization ka hi ek variation hai, par isme variance ko $\sqrt{2 / n}$ se scale kiya jaata hai.

* **Best For:** **ReLU** activation function ke saath.
* **Benefit:** ReLU me aadhe neurons turn off (0) ho jaate hain, toh yeh formula unhe "dead neurons" banne se bachata hai aur training fast rakhta hai.

---

### **3. Activation Functions aur Initialization Ka Match**

* **Sigmoid Activation:** Isme vanishing gradient ka khatra sabse zyaada hota hai. Iske liye **Xavier Initialization** sabse best hai.
* **Tanh Activation:** Sigmoid ki tarah isme bhi vanishing gradient ho sakta hai. Iske saath bhi **Xavier Initialization** use karna chahiye.
* **ReLU Activation:** ReLU non-saturating hota hai, lekin bade weights hone par gradients explode ho sakte hain. Iske liye **He Initialization** sabse badhiya kaam karta hai.

---

### **Quick Cheat-Sheet (Konsa Kab Use Karein?)**

* **ReLU Activation Use Kar Rahe Ho?** $\rightarrow$ **He Initialization** choose karo.
* **Tanh ya Sigmoid Use Kar Rahe Ho?** $\rightarrow$ **Xavier Initialization** choose karo.
* **Zero Initialization?** $\rightarrow$ Kabhi mat use karo!
