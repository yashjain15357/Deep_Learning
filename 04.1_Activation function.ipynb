### **Which Activation Function to use when?**

Activation functions choose karte time, problem ki category aur neural network ke layer (Hidden Layer vs. Output Layer) par depend karta hai.

**1. Hidden Layers ke liye (Best Options)**

* **ReLU (Rectified Linear Unit):** Standard choice for hidden layers. High computational efficiency aur faster convergence milti hai.
* **Leaky ReLU / ELU:** Agar ReLU ke sath *Dying ReLU problem* (jab neurons negative values ki wajah se 0 par stuck ho jate hain) aaye, toh ise use karte hain.
* **Tanh (Hyperbolic Tangent):** Zero-centered output $(-1 \text{ to } +1)$ chahiye ho toh useful hai, par deep networks me *Vanishing Gradient* ho sakta hai.

---

**2. Output Layer ke liye (Problem-Specific)**

* **Binary Classification (2 Classes):** **Sigmoid**
* Output range $(0 \text{ to } 1)$ hoti hai, jo probability ki tarah treat hoti hai (e.g., Cat vs. Dog, Spam vs. Not Spam).


* **Multi-Class Classification (3+ Classes):** **Softmax**
* Saare classes ke raw scores ko probabilities me convert karta hai jinka sum 1 hota hai (e.g., Digit recognition 0-9).


* **Regression (Continuous Output):** **Linear / Identity Activation (No Activation)**
* Jab continuous real numbers predict karne hon (e.g., House Price prediction, Temperature prediction).



---

**Quick Summary Table**

| Layer / Problem Type | Recommended Activation | Primary Reason |
| --- | --- | --- |
| **Hidden Layers** | **ReLU** | Fast execution, default standard choice |
| **Hidden Layers (Dying ReLU Issue)** | **Leaky ReLU / ELU** | Negative values par small gradient allow karta hai |
| **Output Layer (Binary Class)** | **Sigmoid** | Output ko 0-1 probability range me map karta hai |
| **Output Layer (Multi-Class)** | **Softmax** | Multiple classes ke probability scores sum to 1 karte hain |
| **Output Layer (Regression)** | **Linear** | Unbounded numeric values predict karne ke liye |
