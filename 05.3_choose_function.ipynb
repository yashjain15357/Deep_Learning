## **Which Activation and Loss Function use when?**

| Problem Statement | Hidden Layer Activation | Output Layer Activation | Loss Function | Recommended Activation Function (Summary) |
| :--- | :--- | :--- | :--- | :--- |
| **Binary Classification**<br>*(e.g., Email Spam vs Not Spam, Tumor Malignant/Benign)* | **ReLU** / **LeakyReLU** | **Sigmoid** | **Binary Cross-Entropy** | **Hidden:** ReLU / LeakyReLU<br>**Output:** Sigmoid |
| **Multi-Class Classification**<br>*(Targets as One-Hot Vectors, e.g., `[0, 1, 0]`)* | **ReLU** / **GELU** | **Softmax** | **Categorical Crossentropy** | **Hidden:** ReLU / GELU<br>**Output:** Softmax |
| **Multi-Class Classification**<br>*(Targets as Integers, e.g., Class `2`)* | **ReLU** / **GELU** | **Softmax** | **Sparse Categorical Crossentropy** | **Hidden:** ReLU / GELU<br>**Output:** Softmax |
| **Multi-Label Classification**<br>*(Multiple valid tags per image/text, e.g., `[1, 0, 1]`)* | **ReLU** / **LeakyReLU** | **Sigmoid** *(per node)* | **Binary Cross-Entropy** | **Hidden:** ReLU / LeakyReLU<br>**Output:** Sigmoid |
| **Regression (Continuous Values)**<br>*(e.g., Predicting House Prices, Temperature)* | **ReLU** / **GELU** | **Linear** *(None)* | **Mean Squared Error (MSE)** / **MAE** | **Hidden:** ReLU / GELU<br>**Output:** Linear |
