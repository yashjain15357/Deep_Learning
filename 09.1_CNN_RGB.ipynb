CNN (Convolutional Neural Network) jab kisi **RGB (Color) Image** par kaam karta hai, toh wo ek structured pipeline ke zariye features extract karta hai aur final classification karta hai.

RGB image mein 3 channels hote hain: **Red, Green, aur Blue**. Isiliye, 2D black-and-white image ki jagah ek RGB image ko computer ek **3D Tensor** $(\text{Height} \times \text{Width} \times 3)$ ke roop mein dekhta hai.

Ek end-to-end example ke saath dekhte hain ki CNN RGB image par step-by-step kaise kaam karta hai:

---

### Example Scenario

* **Input Image:** A color image of size **$6 \times 6$ pixels**.
* **Channels:** $3$ (Red, Green, Blue).
* **Dimensions:** $(6 \times 6 \times 3)$.

---

### Step 1: Input Layer (3D Tensor Representation)

Computer ke liye RGB image teen alag-alag 2D matrices ka stack hoti hai:

1. **Red Channel Matrix:** Pixel values ($0$ to $255$) jo red intensity batate hain.
2. **Green Channel Matrix:** Pixel values green intensity ke liye.
3. **Blue Channel Matrix:** Pixel values blue intensity ke liye.

$$\text{Input Shape} = (H, W, C) = (6, 6, 3)$$

---

### Step 2: Convolutional Layer (Feature Extraction)

Convolutional layer ka main kaam image se features (edges, textures, colors, shapes) extract karna hota hai.

#### 1. Filter (Kernel) Setup

* RGB image par process karne ke liye, **Filter ka Depth hamesha Input Channels ke barabar ($3$) hota hai**.
* Agar hum $3 \times 3$ size ka filter use kar rahe hain, toh filter dimensions hungi: **$(3 \times 3 \times 3)$**.
* Iska matlab filter ke paas bhi 3 slices hote hain: ek Red channel ke liye, ek Green ke liye, aur ek Blue ke liye.

#### 2. Sliding & Element-wise Multiplication

Filter image ke top-left corner par baithta hai aur sliding window mechanism apply hota hai:

1. Filter ka **Red slice**, image ke **Red channel** ke $3 \times 3$ region se multiply hota hai.
2. Filter ka **Green slice**, image ke **Green channel** ke $3 \times 3$ region se multiply hota hai.
3. Filter ka **Blue slice**, image ke **Blue channel** ke $3 \times 3$ region se multiply hota hai.
4. Teeno channels ke results ko add kiya jata hai aur ek single **Bias ($b$)** add hota hai:

$$\text{Output Value} = \sum (\text{Red}_{\text{conv}}) + \sum (\text{Green}_{\text{conv}}) + \sum (\text{Blue}_{\text{conv}}) + \text{Bias}$$

#### 3. Output Feature Map

* Har ek filter pure 3D RGB input ko multiply-add karke **1 Single 2D Feature Map** banata hai.
* Agar hum **$K$ filters** (jaise 16 ya 32 filters) use karte hain, toh 16 ya 32 alag-alag 2D feature maps milte hain.
* **Output Shape Formula:**

$$\text{Output Width/Height} = \frac{W - F + 2P}{S} + 1$$



*(Jahan $W$=Input Size, $F$=Filter Size, $P$=Padding, $S$=Stride)*
* **Example calculation:**
* Input: $6 \times 6 \times 3$, Filter: $3 \times 3 \times 3$, Stride: $1$, Padding: $0$.
* Feature Map Size: $\frac{6 - 3 + 0}{1} + 1 = 4 \times 4$.
* Agar hum **$16$ filters** lagate hain, toh output tensor banega: **$(4 \times 4 \times 16)$**.



---

### Step 3: Activation Function (ReLU)

Convolution ke baad non-linearity introduce karne ke liye **ReLU (Rectified Linear Unit)** activation function lagaya jata hai.

* **Formula:** $f(x) = \max(0, x)$
* Iska kaam saari negative values ko $0$ kar dena hota hai aur positive values ko unchanged chhodna hota hai, jisse model complex non-linear patterns ko learn kar sake.

---

### Step 4: Pooling Layer (Dimensionality Reduction)

Pooling layer image ki height aur width ko reduce karti hai taaki computational complexity kam ho sake, jabki important spatial features retain rahain.

* **Max Pooling Example:** $2 \times 2$ window aur Stride = $2$.
* Output tensor $(4 \times 4 \times 16)$ par jab $2 \times 2$ Max Pooling apply hogi:
* Height aur Width aadhi ho jayegi ($4 \to 2$).
* Channels same rahoge ($16$).


* **New Dimensions:** **$(2 \times 2 \times 16)$**.

---

### Step 5: Stacking Multiple Conv + Pooling Layers

Deep CNNs mein yeh cycle repeat hoti hai:

1. **Initial Layers:** Low-level features extract karti hain (jaise edges, color contrasts).
2. **Deeper Layers:** High-level abstract features compute karti hain (jaise eyes, wheels, faces, object parts).

---

### Step 6: Flattening

Jab feature extraction complete ho jata hai, tab final 3D feature map ko 1D vector mein convert kiya jata hai taaki isse Fully Connected Layer mein bheja ja sake.

* Output Tensor Shape: $(2 \times 2 \times 16)$
* **Flattened 1D Array Size:** $2 \times 2 \times 16 = \mathbf{64 \text{ values}}$.

---

### Step 7: Fully Connected (FC) Layer & Output

1. **Dense Layers:** Flattened $64$-length vector ko fully connected neural network layers mein paas kiya jata hai jo high-level reasoning/classification handle karti hain.
2. **Softmax Output Layer:** Final output layer mein har class ke liye probability score generate hota hai using **Softmax** activation:
* **Cat:** $0.85$ ($85\%$)
* **Dog:** $0.10$ ($10\%$)
* **Car:** $0.05$ ($5\%$)



---

### Key Summary Table for RGB Processing

| Stage | Input Shape Example | Operation / Mechanism | Output Shape Example |
| --- | --- | --- | --- |
| **Input** | - | RGB Image Load (3 Channels) | $6 \times 6 \times 3$ |
| **Conv1 (16 Filters of $3\times3\times3$)** | $6 \times 6 \times 3$ | 3D Convolution across all 3 channels + Bias + ReLU | $4 \times 4 \times 16$ |
| **Max Pooling ($2\times2$)** | $4 \times 4 \times 16$ | Downsampling (Takes max value in $2\times2$ patch) | $2 \times 2 \times 16$ |
| **Flattening** | $2 \times 2 \times 16$ | Reshaping 3D tensor to 1D vector | $1 \times 64$ vector |
| **Fully Connected Layer** | $1 \times 64$ vector | Weighted sum + Classification reasoning | $1 \times N$ classes |
| **Softmax Output** | Dense Output | Calculates class probabilities (Sum = 1) | Probabilities per class |
